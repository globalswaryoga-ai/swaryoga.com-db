```python
# ==============================================================================
# ⚡ STEP 4: FAST NEURAL CHARACTER BLEND (30s @ 1080p)
# ==============================================================================
print("🎬 Initializing Neural Synthesis Engine...")
app = FaceAnalysis(name='buffalo_l', providers=['CUDAExecutionProvider', 'CPUExecutionProvider'])
app.prepare(ctx_id=0, det_size=(640, 640))
synthesizer = insightface.model_zoo.get_model('models/neural_adapter.bin', providers=['CUDAExecutionProvider', 'CPUExecutionProvider'])

master_src = cv2.imread('inputs/mohan_dna.jpg')
master_identity = app.get(master_src)[0]

cap = cv2.VideoCapture('inputs/target_template.mp4')
fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
w, h = int(cap.get(3)), int(cap.get(4))
total_f = min(int(cap.get(cv2.CAP_PROP_FRAME_COUNT)), 900)  # 30 seconds

raw_video = 'outputs/raw_render.mp4'
out = cv2.VideoWriter(raw_video, cv2.VideoWriter_fourcc(*'mp4v'), fps, (w, h))

print(f"🔥 Rendering {total_f} frames ({w}x{h} @ {fps} FPS) on Kaggle GPU...")
start_t = time.time()
for _ in tqdm(range(total_f)):
    ret, frame = cap.read()
    if not ret: break
    detected = app.get(frame)
    if detected:
        frame = synthesizer.get(frame, detected[0], master_identity, paste_back=True)
    out.write(frame)
cap.release()
out.release()
elapsed = time.time() - start_t
print(f"🎉 Complete in {elapsed:.1f}s ({total_f/elapsed:.1f} FPS)!")
```
